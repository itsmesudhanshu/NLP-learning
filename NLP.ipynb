LowerCasing

```python
def lowercase(text):
  return text.lower()

p="This is a MIXED case Text"
lowercased_text= lowercase(p)
print(lowercased_text)
```

```python
def remov_stop(j,k):
  words=j.split()
  filtered_word=[words for l in words if l not in k]
  return ' '.join(filtered_word)


text= "My name is sudhanshu chouhan and i want to be PM of this country"
stop_word={"is","and", "the"}
filtered_text= remov_stop(text,stop_word)
print("text after stop word removal", filtered_text)
```

#Stemming demo

```python
def stem(word):
  suffixes=["ing","ied","en"]
  for i in suffixes:
    if word.endswith(i):
      return word[:-len(i)-1]
x= ["running", "studied","fasten"]
y= [stem(t) for t in x]
print(y)
```

#Lemmatization Demo

```python
def lemma(word):
  base={"running":"run","studied":"study","better":"good"}
  if word in base:
    return base[word]
words=["running", "studied", "better"]
lem= [lemma(j) for j in words]
print(lem)
```
